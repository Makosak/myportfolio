## Subset Data

A common task for us will be subsetting data, or taking a component of the bigger dataset. 

Let's say we want a data frame with only information about the type of airport, name, it's abbreviation, and code. We'll give it a new name.

```python
airports1 = airports[["name","type","abbrev","natlscale"]]
```

```python
airports1.head()
```

### How many categories?
We can see that there are a few different types of airports in the dataset. How many types? We can use `describe` for the basic information about the data.

```python
print(airports1['type'].describe())
```

### Which categories?
What are the specific data cateogries that we have for airports? We can use the `unique` function for that info.

```python
print(airports1['type'].unique())
```

***

## Filter Data
Sometimes we may want to set conditions to filter the data. That could mean, for example: subsetting to only small airports, or only including aiports that have a low national scale. 

### Select By Attribute
In the non-spatial world, we are generally limited to selecting by attribute. That means we can only use the data we see in the table, working the attributes of data. 

Let's start with an example where filter by a category, like only grabbing airports that are small.

```python
smallAirports = airports1[airports1["type"] == 'small']
smallAirports.head()
```

Note that we use two equal signs to correspond to the boolean logic "is equal to" -- this may be new to you! In the result, we only have 2 small airports in a our dataset.

Next, let's grab airpots that have a national scale that is above 9.0.

```python
hiScaleAiports = airports1[airports1["natlscale"] < 9]
hiScaleAiports.head()
```
